Import and clean spotify data

```python
import os
import json
import pandas as pd

dir = "../data/raw"
```

```python
  # counts num of objects in json
def count_objects(filepath):
    with open(filepath, 'r', encoding="utf-8") as f:
        data = json.load(f)
    if isinstance(data, list):
        return len(data)
    
count = 0    

# iterate through files
for file in os.listdir(dir):
    filename = os.fsdecode(file)
    if filename.endswith(".json"): 
        file_path = os.path.join(dir, filename)
        print(filename)
        objs = count_objects(file_path)
        print(objs)
        count += objs
        continue
    else:
        continue

# total songs
print(count)
```

Check out variables

```python
filepath = f'{dir}/Streaming_History_Audio_2025_3.json'

with open(filepath, 'r', encoding="utf-8") as f:
        data = json.load(f)

# get the variable names
print(data[0].keys())
# get variable values
print(data[0].values())
```

Convert to dataframe

```python
# convert to df first
df = pd.DataFrame(data)
print(df)
```

Select certain variables

```python
df = df[['ts', 'ms_played', 'master_metadata_track_name', 'master_metadata_album_artist_name',
         'master_metadata_album_album_name','spotify_track_uri']]
```

Clean data

```python
# look for anything unusual
print(df.isna().sum())
```

```python
# Some rows are missing important variables, non-songs that got through?
print(df[df['master_metadata_track_name'].isna()])

# drop them
df_clean = df.dropna(subset=['master_metadata_track_name'])
```

```python
# check again, this time we're good!
print(df_clean.isna().sum())
```

Save to csv

```python
print(os.getcwd())
df.to_csv('../data/clean/data_25')
```
