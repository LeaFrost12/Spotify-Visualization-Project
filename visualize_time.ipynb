```python
import pandas as pd
import plotly.express as px
from src.clean import load_all_data
```

Load in data as df

```python
df = load_all_data()
```

### 1. Line plot of listening over time

No aggregation

```python
fig = px.line(df, x="ts_local", y="ms_played", title='Life expectancy in Canada')
fig.show()
```

aggregate by day

```python
# convert to datetime
df['ts_local'] = pd.to_datetime(df['ts_local'], errors='coerce', utc=True)

#group by date
daily = df.groupby(df['ts_local'].dt.date)['ms_played'].sum()

print(daily)

```

```python
fig = px.line(daily, x=daily.index, y="ms_played", title='Life expectancy in Canada')
fig.show()
```

Aggregate by week

```python
# calculate the mins
daily_min = daily / (1000 * 60)
# convert to df
daily_df = daily_min.reset_index(name='min_played')
# aggregate
daily_df['7d_avg'] = daily_df['min_played'].rolling(7, min_periods=1).mean()
print(daily_df)

# ensure datetime
daily_df['ts_local'] = pd.to_datetime(daily_df['ts_local'])

# filter out rows before cutoff
cutoff = "2021-05-01"
daily_df = daily_df[daily_df['ts_local'] >= cutoff]

fig = px.line(daily_df, x='ts_local', y='min_played', title='Daily listening (min)')
fig.update_traces(opacity=0.2)
fig.add_scatter(x=daily_df['ts_local'], y=daily_df['7d_avg'], mode='lines', name='7-day avg')
fig.update_xaxes(title='Date')
fig.update_yaxes(title='Minutes listened')
fig.show()

```

Aggregate by month

```python
# calculate the mins
daily_min = daily / (1000 * 60)
# convert to df
daily_df = daily_min.reset_index(name='min_played')
# aggregate
daily_df['30d_avg'] = daily_df['min_played'].rolling(30, min_periods=1).mean()
print(daily_df)

# ensure datetime
daily_df['ts_local'] = pd.to_datetime(daily_df['ts_local'])

# filter out rows before cutoff
cutoff = "2021-05-01"
daily_df = daily_df[daily_df['ts_local'] >= cutoff]

fig = px.line(daily_df, x='ts_local', y='min_played', title='Daily listening (min)')
fig.update_traces(opacity=0.2)
fig.add_scatter(x=daily_df['ts_local'], y=daily_df['30d_avg'], mode='lines', name='30-day avg')
fig.update_xaxes(title='Date')
fig.update_yaxes(title='Minutes listened')
fig.show()
```
